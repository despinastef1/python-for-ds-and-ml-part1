```python
from os import system
import random

def start_game():
    system("clear")
    print("Welcome to my Math game!")
    x = input('''What would you like to practice on? 
    (1 - addition, 2 - subtraction, 3 -multiplication, 4 - division)''')
    points = 0
    index = 0
    while index < 3:
        if x == '1':
            print("You picked addition")
            num1 = random.randint(1,100)
            num2 = random.randint(1,100)
            sum = num1 + num2
            your_answer = input(f"{num1} + {num2} = ")
            if int(your_answer) == sum:
                print("That is correct!")
                points += 1
            else:
                print("That is incorrect!")
            index += 1
        elif x == '2':
            print("You picked subtraction")
            num1 = random.randint(1,100)
            num2 = random.randint(1,100)
            diff = num1 - num2
            your_answer = input(f"{num1} - {num2} = ")
            if int(your_answer) == diff:
                print("That is correct!")
                points += 1
            else:
                print("That is incorrect!")
            index += 1
        elif x == '3':
            print("You picked multiplication")
            num1 = random.randint(1,100)
            num2 = random.randint(1,100)
            prod = num1 * num2
            your_answer = input(f"{num1} * {num2} = ")
            if int(your_answer) == prod:
                print("That is correct!")
                points += 1
            else:
                print("That is incorrect!")
            index += 1
        elif x == '4':
            print("You picked division")
            num1 = random.randint(1,100)
            num2 = random.randint(1,100)
            div = num1/num2
            your_answer = input(f"{num1} / {num2} = ")
            if int(your_answer) == div:
                print("That is correct!")
                points += 1
            else:
                print("That is incorrect!")
            index += 1
        else: 
            print("Pick a value between 1 and 4!")
            break
    print(f"You got {points} points!")
    
start_game()
```
