```python
import numpy as np
import pandas as pd

from pandas import DataFrame
```

```python
# create a df object
numbers_df = DataFrame(np.arange(0,90,3).reshape(10,3), 
index = ['row 1','row 2','row 3','row 4','row 5','row 6','row 7','row 8','row 9','row 10'], 
columns = ['column 1','column 2', 'column 3'])
```

```python
numbers_df
```

```python
#indexing and slicing

#simple indexing
numbers_df.iloc[0,1]
```

```python
#replace element [0,1]
numbers_df.iloc[0,1]=20
numbers_df
```

```python
# use fancy indexing
numbers_df.iloc[[1,2,4],[1,2]]
```

```python
#boolean indexing / boolean mask
mask = numbers_df > 30
mask
```

```python
numbers_df[mask]
```

```python
numbers_df[numbers_df>30] = 0
numbers_df
```

```python
#slicing values from the data frame
numbers_df.iloc[2:6,1:3]
```
