```python
import numpy as np
import pandas as pd

from pandas import DataFrame
```

```python
# filling missing values using fillna(), replace(), and interpolate()
data = {'names':['steve','john','richard','sarah','randy','michael','julie'],
       'age':[20,22,20,21,24,23,22],
       'gender':['Male','Male','Male','Female','Male','Male','Female'],
       'rank':[2,1,4,5,3,7,6]  
       }
ranking_df = DataFrame(data)
ranking_df.iloc[2:5,1] = np.nan
ranking_df.iloc[3:6,3] = np.nan
ranking_df.iloc[3,:] = np.nan
ranking_df
```

```python
# handling missing values
ranking_df.isnull()
```

```python
ranking_df.notnull()
```

```python
bool_series = pd.isnull(ranking_df['age'])
ranking_df[bool_series]
```

```python
# fill missing values in a data frame

#fill all missing with 0
ranking_df.fillna(0)
```

```python
# Use ffill() to fill each missing value with the preceding value in the same column:
ranking_df.ffill()
```

```python
# Use ffill() to fill each missing value with the next value in the same column:
ranking_df.bfill()
```

```python
# fill with interpolated values
numeric_cols = ranking_df.select_dtypes(include='number').columns
ranking_df[numeric_cols] = ranking_df[numeric_cols].interpolate(method='linear')
ranking_df
```

```python
# drop rows and columns that have NaN
ranking_df.dropna()
```

```python
#drop all rows and columns that contain only nas
ranking_df.dropna(how='all')
```

```python
#drop columns that contain at least one missing values
ranking_df.dropna(axis = 1)
```

```python
#drop rows that contain at least one missing values
ranking_df.dropna(axis = 0)
```
