```python
import pandas as pd

df1 = pd.DataFrame({
    'Name': ['Anna', 'Ben'],
    'Age': [25, 30]
})

df2 = pd.DataFrame({
    'Name': ['Chris', 'Diana'],
    'Age': [28, 35]
})
```
