```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sb
from matplotlib import rcParams

import sklearn
from sklearn import preprocessing
from sklearn.preprocessing import scale
```

```python
%matplotlib inline
rcParams['figure.figsize'] = 8,4
sb.set_style('whitegrid')
```

```python
address = r"C:\Users\despi\OneDrive\Desktop\Python for DS and ML Part 1 - older version\data\mtcars.csv"
cars = pd.read_csv(address)
cars.columns = ['car_names', 'mpg',	'cyl',	'disp',	'hp',	'drat',	'wt',	'qsec',	'vs',	'am',	'gear',	'carb']
```

```python
#normalizing and transforming features
mpg = cars.mpg
plt.plot(mpg)
```

```python
cars[['mpg']].describe()
```

```python
mpg.describe()
```

```python
mpg_matrix = mpg.values.reshape(-1,1) #shapes mpg as a 1-column matrix
```

```python
scaled = preprocessing.MinMaxScaler()
scaled_mpg = scaled.fit_transform(mpg_matrix)
plt.plot(scaled_mpg)
```

```python
scaled = preprocessing.MinMaxScaler(feature_range=(0,10))
scaled_mpg = scaled.fit_transform(mpg_matrix)
plt.plot(scaled_mpg)
```

```python
standardized_mpg = scale(mpg, axis=0, with_mean=False, with_std=False)
plt.plot(standardized_mpg)
```

```python
standardized_mpg = scale(mpg)
plt.plot(standardized_mpg)
```
