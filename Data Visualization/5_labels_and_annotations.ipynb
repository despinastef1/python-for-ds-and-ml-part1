```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from numpy.random import randn
from pandas import Series, DataFrame
from matplotlib import rcParams
```

```python
%matplotlib inline
#this will print plots that are 5 inches wide and 4 inches tall
rcParams['figure.figsize'] = 12,4

sns.set_style('whitegrid')
```

```python
#labeling plot features
#the functional method
x = range(1,10)
y = [1,2,3,4,0.5,4,3,2,1]

plt.bar(x,y)
plt.xlabel('your x-axis label')
plt.ylabel('your y-axis label')
```

```python
z = [1,2,3,4,0.5]
veh_type = ['bicycle', ' motorbikr', 'car', 'van','stroller']
plt.pie(z, labels = veh_type)
```

```python
# the object-oriented method
address = r'C:\Users\despi\OneDrive\Desktop\Python for DS and ML Part 1\data\mtcars.csv'
cars = pd.read_csv(address)

cars.columns = ['car_names', 'mpg','cyl','disp','hp','drat','wt','qsec','vs','am','gear','carb']
mpg = cars['mpg']

fig = plt.figure()
ax = fig.add_axes([.1,.1,1,1])

mpg.plot()

ax.set_xticks(range(32))
ax.set_xticklabels(cars.car_names, rotation = 60, fontsize = 'medium')
ax.set_title('Miles Per Gallon of Cars in mtcars')

ax.set_xlabel('car names')
ax.set_ylabel('miles/gal')
```

```python
#add a legend
#call the legend method .legend(label,loc)
#https://matplotlib.org/api/pyplot_api.html#matplotlib.pyplot.legend

#functional method
plt.pie(z)
plt.legend(veh_type,loc = 'best')
plt.show()
```

```python
#object-oriented method
fig = plt.figure()
ax = fig.add_axes([.1,.1,1,1])

mpg.plot()

ax.set_xticks(range(32))
ax.set_xticklabels(cars.car_names, rotation = 60, fontsize = 'medium')
ax.set_title('Miles Per Gallon of Cars in mtcars')

ax.set_xlabel('car names')
ax.set_ylabel('miles/gal')

ax.legend(loc='best')
```

```python
fig = plt.figure()
ax = fig.add_axes([.1,.1,1,1])

mpg.plot()

ax.set_xticks(range(32))
ax.set_xticklabels(cars.car_names, rotation = 60, fontsize = 'medium')
ax.set_title('Miles Per Gallon of Cars in mtcars')

ax.set_xlabel('car names')
ax.set_ylabel('miles/gal')

ax.set_ylim([0,45])

ax.annotate('Toyota Corolla', xy=(19,33.9), xytext=(21,35),arrowprops = dict(facecolor='black', shrink=0.05))

ax.legend(loc='best')
```
