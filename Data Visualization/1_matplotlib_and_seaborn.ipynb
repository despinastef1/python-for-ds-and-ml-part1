```python
import matplotlib.pyplot as plt
import seaborn as sns

from pandas import DataFrame
```

```python
data = {'names':['steve','john','richard','sarah','randy','michael','julie'],
       'age':[20,22,20,21,24,23,22],
       'gender':['Male','Male','Male','Female','Male','Male','Female'],
       'rank':[2,1,4,5,3,7,6]  
       }
df = DataFrame(data)
df
```

```python
plt.bar(df['names'], df['age'])
plt.xlabel('Names')
plt.ylabel('Age')
plt.title('Comparing Ages')
plt.show()
```

```python
plot=sns.barplot(data=df,x='names', y='age')
plot.set_title('Comparing Ages')
plt.show()
```

```python
plt.plot(df['names'],df['age'])
plt.xlabel('Names')
plt.ylabel('Age')
plt.title('Comparing Ages')
plt.show()
```

```python
plot = sns.lineplot(data=df, x='names', y='age')
plot.set_title('Comparing Ages')
plt.show()
```

```python
plt.pie(df['age'], labels = df['names'])
plt.title('Age Comparison')
plt.show()
```

```python
#pie chart seaborn - there is no function for pie chart in seaborn, but can define a new seaborn color palette and apply to plt
colors = sns.color_palette('pastel')[0:5]
plt.pie(df['age'], labels = df['names'], colors = colors)
plt.show()
```
