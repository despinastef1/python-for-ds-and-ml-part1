```python
actual = [10, 13, 16]
model_pred = [11, 13, 14]

abs_diff = [abs(x - y) for x,y in zip(actual,model_pred)]
mae = sum(abs_diff)/len(abs_diff)
print(mae)
```

```python
import numpy as np

actual = [10, 13, 16]
model_pred = [11, 13, 14]

actual = np.array(actual)
model_pred = np.array(model_pred)

abs_diff = [abs(x - y) for x,y in zip(actual,model_pred)]
mae = sum(abs_diff)/len(abs_diff)
print(mae)
```
