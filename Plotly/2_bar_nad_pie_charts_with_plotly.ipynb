```python
import plotly.graph_objects as go
```

```python
# Define data
data = [go.Bar(
    x=[1, 2, 3, 4, 5, 6, 7, 8, 9],
    y=[1, 2, 3, 4, 0.5, 4, 3, 2, 1]
)]

# Create figure
fig = go.Figure(data=data)

# Customize chart
fig.update_layout(
    title="Basic Bar Chart",
    xaxis_title="X Values",
    yaxis_title="Y Values"
)

# Display chart
fig.show()
```

```python
fig = go.Figure(
    data=[
        go.Pie(
            labels=['bicycle', 'motorcycle', 'car', 'van', 'stroller'],
            values=[1, 2, 3, 4, 0.5]
        )
    ]
)

fig.update_layout(title="Simple Pie Chart")

fig.show()
```
